# Avocados Memorandum Calculations Verification

```python
# Import Packages
import pandas as pd 
```

```python
# Import the data and store
data2025=pd.read_csv('2025.csv',encoding='utf-8')
data2026=pd.read_csv('2026PartYear.csv',encoding='utf-8')
```

```python
# Total Change in U.S. Volume YoY
data2025_32 = data2025[data2025['Period'] <= 32]

volume2025 = data2025_32[data2025_32['Geography'] == 'Total U.S.']['Total Bulk and Bags'].sum()
volume2026 = data2026[data2026['Geography'] == 'Total U.S.']['Total Bulk and Bags'].sum()

volume_change = (volume2026 / volume2025 - 1) * 100

print(round(volume2025 / 1e9, 2))
print(round(volume2026 / 1e9, 2))
print(round(volume_change, 1))
```

```python
# Total U.S. Volume-Weighted ASP ($)
us2025 = data2025_32[data2025_32['Geography'] == 'Total U.S.']
us2026 = data2026[data2026['Geography'] == 'Total U.S.']
asp2025 = (us2025['ASP Current Year'] * us2025['Total Bulk and Bags']).sum() / volume2025
asp2026 = (us2026['ASP Current Year'] * us2026['Total Bulk and Bags']).sum() / volume2026
print(round(asp2025, 2))
print(round(asp2026, 2))
```

```python
# Percentage Change in Total U.S. ASP
asp_change = (asp2026 / asp2025 - 1) * 100
print(round(asp_change, 1))
```

```python
# Change Conventional Volume
conv2025 = us2025[us2025['Type'] == 'Conventional']
conv2026 = us2026[us2026['Type'] == 'Conventional']
conv_volume2025 = conv2025['Total Bulk and Bags'].sum()
conv_volume2026 = conv2026['Total Bulk and Bags'].sum()
print(round(conv_volume2025 / 1e9, 2))
print(round(conv_volume2026 / 1e9, 2))
```

```python
# Percentage Change in Conventional Volume
conv_volume_change = (conv_volume2026 / conv_volume2025 - 1) * 100
print(round(conv_volume_change, 1))
```

```python
# Conventional Volume-Weighted ASP ($)
conv_asp2025 = (conv2025['ASP Current Year'] * conv2025['Total Bulk and Bags']).sum() / conv_volume2025
conv_asp2026 = (conv2026['ASP Current Year'] * conv2026['Total Bulk and Bags']).sum() / conv_volume2026
print(round(conv_asp2025, 2))
print(round(conv_asp2026, 2))
```

```python
# Percentage Change Conventional ASP
conv_asp_change = (conv_asp2026 / conv_asp2025 - 1) * 100
print(round(conv_asp_change, 1))
```

```python
# Conventional Share of Total U.S. Volume
conv_share2025 = conv_volume2025 / volume2025 * 100
conv_share2026 = conv_volume2026 / volume2026 * 100
print(round(conv_share2025, 1))
print(round(conv_share2026, 1))
```

```python
# Organic U.S. Volume
org2025 = us2025[us2025['Type'] == 'Organic']
org2026 = us2026[us2026['Type'] == 'Organic']
org_volume2025 = org2025['Total Bulk and Bags'].sum()
org_volume2026 = org2026['Total Bulk and Bags'].sum()
print(round(org_volume2025 / 1e6, 1))
print(round(org_volume2026 / 1e6, 1))
```

```python
# Percentage Change Organic Volume
org_volume_change = (org_volume2026 / org_volume2025 - 1) * 100
print(round(org_volume_change, 1))
```

```python
# Organic Volume-Weighted ASP
org_asp2025 = (org2025['ASP Current Year'] * org2025['Total Bulk and Bags']).sum() / org_volume2025
org_asp2026 = (org2026['ASP Current Year'] * org2026['Total Bulk and Bags']).sum() / org_volume2026
print(round(org_asp2025, 2))
print(round(org_asp2026, 2))
```

```python
# Percentage Change Organic ASP
org_asp_change = (org_asp2026 / org_asp2025 - 1) * 100
print(round(org_asp_change, 1))
```

```python
# Conventional 4046 Volume Growth
conv4046_2025 = conv2025['4046'].sum()
conv4046_2026 = conv2026['4046'].sum()
conv4046_change = (conv4046_2026 / conv4046_2025 - 1) * 100
print(round(conv4046_2025 / 1e6, 1))
print(round(conv4046_2026 / 1e6, 1))
print(round(conv4046_change, 1))
```

```python
# Conventional 4226 Volume Growth
conv4225_2025 = conv2025['4225'].sum()
conv4225_2026 = conv2026['4225'].sum()
conv4225_change = (conv4225_2026 / conv4225_2025 - 1) * 100
print(round(conv4225_2025 / 1e6, 1))
print(round(conv4225_2026 / 1e6, 1))
print(round(conv4225_change, 1))
```

```python
# Conventional 4770 Volume Growth
conv4770_2025 = conv2025['4770'].sum()
conv4770_2026 = conv2026['4770'].sum()
conv4770_change = (conv4770_2026 / conv4770_2025 - 1) * 100
print(round(conv4770_2025 / 1e6, 1))
print(round(conv4770_2026 / 1e6, 1))
print(round(conv4770_change, 1))
```

```python
# Organic Avocado Type Changes
for avocado in ['4046', '4225', '4770']:
    change = (org2026[avocado].sum() / org2025[avocado].sum() - 1) * 100
    print(avocado, round(change, 1))
```

```python
# Conventional Bagged Volume Growth
conv_bag2025 = conv2025['TotalBagged'].sum()
conv_bag2026 = conv2026['TotalBagged'].sum()
conv_bag_change = (conv_bag2026 / conv_bag2025 - 1) * 100
print(round(conv_bag2025 / 1e6, 1))
print(round(conv_bag2026 / 1e6, 1))
print(round(conv_bag_change, 1))
```

```python
# Conventional Bag Size Growth
for bag in ['Sml Bagged', 'Lrg Bagged', 'X-Lrg Bagged']:
    change = (conv2026[bag].sum() / conv2025[bag].sum() - 1) * 100
    print(bag, round(change, 1))
```

```python
# Total Bagged Share of U.S. Volume
bag_share2025 = us2025['TotalBagged'].sum() / volume2025 * 100
bag_share2026 = us2026['TotalBagged'].sum() / volume2026 * 100
print(round(bag_share2025, 1))
print(round(bag_share2026, 1))
```

```python
# Conventional Volume Growth by Region
regions = ['California', 'Southeast', 'Northeast', 'West',
           'South Central', 'Great Lakes', 'Midsouth']
for region in regions:
    vol2025 = data2025_32[(data2025_32['Geography'] == region) &
                          (data2025_32['Type'] == 'Conventional')]['Total Bulk and Bags'].sum()

    vol2026 = data2026[(data2026['Geography'] == region) &
                       (data2026['Type'] == 'Conventional')]['Total Bulk and Bags'].sum()

    change = (vol2026 / vol2025 - 1) * 100
    print(region,
          round(vol2025 / 1e6, 1),
          round(vol2026 / 1e6, 1),
          round(change, 1))
```

```python
# Organic Volume Growth by Region
org_regions = ['California', 'Northeast', 'West', 'Midsouth',
               'Great Lakes', 'Plains', 'San Francisco']

for region in org_regions:
    vol2025 = data2025_32[(data2025_32['Geography'] == region) &
                          (data2025_32['Type'] == 'Organic')]['Total Bulk and Bags'].sum()

    vol2026 = data2026[(data2026['Geography'] == region) &
                       (data2026['Type'] == 'Organic')]['Total Bulk and Bags'].sum()

    change = (vol2026 / vol2025 - 1) * 100
    print(region, round(change, 1))
```
